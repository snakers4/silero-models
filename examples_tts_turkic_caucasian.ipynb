## Caucasian

### Load model

```python
from silero import silero_tts

model_id = 'v5_caucasian'

device = torch.device('cpu')

model, example_text = silero_tts(language='ru',
                                 speaker=model_id)
model.to(device)  # gpu or cpu
```

```python
# ALTERNATIVELY, load directly, without pip

device = torch.device('cpu')

model, _ = torch.hub.load(repo_or_dir='snakers4/silero-models',
                          model='silero_tts',
                          language='ru',
                          speaker='v5_caucasian',
                          force_reload=True)

model.to(device)  # gpu or cpu
```

### List speakers

```python
sorted(model.speakers)
```

### Example

```python
sample_rate = 48000
speaker = 'kbd_1'

example_text = 'Зэи тхузэӏумыхын хуэдэу зэӏуаща щэхущ ахэр.'

audio = model.apply_tts(text=example_text,
                        speaker=speaker,
                        sample_rate=sample_rate)
print(example_text)
display(Audio(audio, rate=sample_rate))
```

### SSML example

- Supported SSML tags can be found here https://github.com/snakers4/silero-models/wiki/SSML
- Text should start and end with the root tag <speak></speak>;
- Use the `ssml_text` parameter

```python
sample_rate = 48000
speaker = 'oss_0'

ssml_text = '<speak>Уыдӕттыл куы ахъуыды кӕнын,<break time="2000ms"/> уӕд фырадӕргӕй, цы фӕуон, уый нал фӕзонын.</speak>'

audio = model.apply_tts(ssml_text=ssml_text,
                        speaker=speaker,
                        sample_rate=sample_rate)
print(ssml_text)
display(Audio(audio, rate=sample_rate))
```

## Turkic

### Load model

```python
from silero import silero_tts

model_id = 'v5_turkic'

device = torch.device('cpu')

model, example_text = silero_tts(language='ru',
                                 speaker=model_id)
model.to(device)  # gpu or cpu
```

```python
# ALTERNATIVELY, load directly, without pip

device = torch.device('cpu')

model, _ = torch.hub.load(repo_or_dir='snakers4/silero-models',
                          model='silero_tts',
                          language='ru',
                          speaker='v5_turkic',
                          force_reload=True)

model.to(device)  # gpu or cpu
```

### List speakers

```python
sorted(model.speakers)
```

### Example

```python
sample_rate = 48000
speaker = 'tat_3'

example_text = 'Болар барысы да шул кадәр серле һәм акыл җитмәслек катлаулы.'

audio = model.apply_tts(text=example_text,
                        speaker=speaker,
                        sample_rate=sample_rate)
print(example_text)
display(Audio(audio, rate=sample_rate))
```

### SSML example

- Supported SSML tags can be found here https://github.com/snakers4/silero-models/wiki/SSML
- Text should start and end with the root tag <speak></speak>;
- Use the `ssml_text` parameter

```python
sample_rate = 48000
speaker = 'sah_0'

ssml_text = '<speak>Барыта туох эрэ таабырыннаах,<break time="2000ms"/> киһи өйүгэр баппат.</speak>'

audio = model.apply_tts(ssml_text=ssml_text,
                        speaker=speaker,
                        sample_rate=sample_rate)
print(ssml_text)
display(Audio(audio, rate=sample_rate))
```
